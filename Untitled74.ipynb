```python
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency
from scipy.stats import norm

file_name = "Cleaned_Sample_Dataset.xlsx"

df = pd.read_excel(file_name)

df["Experiment_Date"] = pd.to_datetime(
    df["Experiment_Date"],
    errors="coerce"
)

summary = []

for variant in df["Variant"].unique():

    temp = df[df["Variant"] == variant]

    users = temp["User_ID"].nunique()

    conversions = temp["Conversion"].sum()

    conversion_rate = (
        conversions / users
        if users > 0
        else 0
    )

    revenue = temp["Revenue"].sum()

    revenue_per_user = (
        revenue / users
        if users > 0
        else 0
    )

    summary.append([
        variant,
        users,
        conversions,
        conversion_rate,
        revenue,
        revenue_per_user
    ])

result = pd.DataFrame(
    summary,
    columns=[
        "Variant",
        "Users",
        "Conversions",
        "Conversion_Rate",
        "Revenue",
        "Revenue_Per_User"
    ]
)

control = result[
    result["Variant"] == "Control"
]

treatment = result[
    result["Variant"] == "Treatment"
]

if len(control) > 0 and len(treatment) > 0:

    control_users = control.iloc[0]["Users"]

    treatment_users = treatment.iloc[0]["Users"]

    control_conversions = (
        control.iloc[0]["Conversions"]
    )

    treatment_conversions = (
        treatment.iloc[0]["Conversions"]
    )

    control_rate = (
        control_conversions /
        control_users
    )

    treatment_rate = (
        treatment_conversions /
        treatment_users
    )

    conversion_lift = (
        (treatment_rate - control_rate)
        / control_rate
        * 100
        if control_rate != 0
        else 0
    )

    pooled_rate = (
        control_conversions +
        treatment_conversions
    ) / (
        control_users +
        treatment_users
    )

    standard_error = np.sqrt(
        pooled_rate *
        (1 - pooled_rate) *
        (
            1 / control_users +
            1 / treatment_users
        )
    )

    if standard_error != 0:

        z_score = (
            treatment_rate -
            control_rate
        ) / standard_error

    else:

        z_score = 0

    p_value = (
        2 *
        (
            1 -
            norm.cdf(abs(z_score))
        )
    )

    contingency_table = pd.crosstab(
        df["Variant"],
        df["Conversion"]
    )

    chi_square, srm_p_value, _, _ = (
        chi2_contingency(contingency_table)
    )

    confidence_error = 1.96 * np.sqrt(
        (
            treatment_rate *
            (1 - treatment_rate)
            / treatment_users
        )
        +
        (
            control_rate *
            (1 - control_rate)
            / control_users
        )
    )

    ci_lower = (
        treatment_rate -
        control_rate -
        confidence_error
    )

    ci_upper = (
        treatment_rate -
        control_rate +
        confidence_error
    )

    treatment_user_share = (
        treatment_users /
        (
            control_users +
            treatment_users
        )
    )

    if p_value < 0.05:

        decision = "Significant"

    else:

        decision = "Not Significant"

    metrics = pd.DataFrame({

        "Metric": [

            "Control Conversion Rate",

            "Treatment Conversion Rate",

            "Conversion Lift (%)",

            "Z Score",

            "P Value",

            "Confidence Interval Lower",

            "Confidence Interval Upper",

            "SRM P Value",

            "Treatment User Share",

            "Decision"

        ],

        "Value": [

            control_rate,

            treatment_rate,

            conversion_lift,

            z_score,

            p_value,

            ci_lower,

            ci_upper,

            srm_p_value,

            treatment_user_share,

            decision

        ]

    })

    variant_result = result.copy()

    variant_result["Section"] = (
        "Variant Summary"
    )

    metrics["Section"] = (
        "Experiment Metrics"
    )

    final_output = pd.concat(
        [
            variant_result,
            metrics
        ],
        ignore_index=True
    )

else:

    final_output = result.copy()

    final_output["Section"] = (
        "Variant Summary"
    )

final_output.to_csv(
    "Python_Analysis_Result.csv",
    index=False
)

print("Analysis completed successfully.")

print("\nOne output file created:")
print("Python_Analysis_Result.csv")

print("\nFinal Analysis Result:")
print(final_output)
```
